```python
import pandas as pd
import numpy as np
import plotly.express as px
from sklearn.linear_model import LinearRegression
from sklearn.compose import ColumnTransformer
from sklearn.preprocessing import OneHotEncoder
```

```python
original_data = pd.read_csv("Loan_default.csv")
```

```python
original_data.head()
```

```python
print(original_data.dtypes)
print(len(original_data))
```

```python
#Just numerical
y=original_data['CreditScore']
x=original_data[['Age',
                 'Income',
                 'MonthsEmployed',
                 'NumCreditLines',
                 'InterestRate',
                 'LoanTerm',
                 'DTIRatio']]
model1=LinearRegression()
model1.fit(x,y)
```

```python
model1.coef_
```

```python
model1.score(x, y)
```

```python
encoder = OneHotEncoder(sparse_output=False,handle_unknown='ignore')
```

```python
cateogorical = ['Education',
                'EmploymentType',
                'MaritalStatus',
                'HasMortgage',
                'HasDependents',
                'LoanPurpose',
                'HasCoSigner',
                'Default']
```

```python
encode_var = encoder.fit_transform(original_data[cateogorical])
encode_var_df = pd.DataFrame(encode_var,columns=encoder.get_feature_names_out(cateogorical))
encode_var_df
```

```python
final = pd.concat([x,encode_var_df],axis=1)
final
```

```python
y=original_data['CreditScore']
x=final[list(final.columns)]
model2=LinearRegression()

model2.fit(x,y)
```

```python
print(model2.coef_)
print(model2.score(x, y))
```

```python
fig = px.scatter(
    original_data,
    x="Age",
    y="LoanAmount",
    title="Loan Amount vs. Credit Score"
)

fig.show()
```
