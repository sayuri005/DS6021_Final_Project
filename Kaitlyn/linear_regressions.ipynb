### Convert Binary Variables to 0 and 1

```python
df[['HasMortgage', 'HasDependents', 'HasCoSigner']] = df[['HasMortgage', 'HasDependents', 'HasCoSigner']].replace({'Yes': 1, 'No': 0}).astype(int)
df
```

Dropping LoanID because I do not think it is needed

```python
df.drop(columns='LoanID', inplace = True)
df
```

Getting the numeric and the categorical columns into their respective lists
---

```python
def get_numerics_cats(df, col_names):
    numeric = []
    categorical = []
    for col_name in col_names:
        if(df[col_name].dtype != 'str'):
            numeric.append(col_name)
        else: 
            categorical.append(col_name)
    return numeric, categorical
```

```python
col_names = df.columns
numeric, categorical = get_numerics_cats(df, col_names)


print(numeric) # NUMERIC CURRENTLY HAS DEFAULT
print(categorical)
```

Linear Regressions
---
I think for predicting a numeric variable and not default, will look at 
- LoanAmount
- InterestRate
- CreditScore?

```python
# sample for bootstrapping
df_sample = (df.sample(n=3000, random_state=6021).reset_index(drop=True))
default_col = df_sample['Default'] # saving and dropping default to avoid data leaks
df_sample.drop(columns='Default', inplace=True)
df_sample
```

Trying LoanAmount at Y
---

```python
y = df_sample['LoanAmount'] # predicting Loan Amount first
df_sample_1 = df_sample.copy()
df_sample_1.drop(columns='LoanAmount', inplace=True)
n = len(df_sample_1)
df_sample_1
```

```python
# drop the columns that we are predicting/cause data leak
numeric_1 = numeric.copy()
numeric_1.remove('LoanAmount')
numeric_1.remove('Default')
print(numeric_1)
```

```python
df_sample[numeric_1].describe()
```

SHOULD NUMCREDITLINES, LOANTERM BE TREATED AS CATEGORICALS?

Trying NumCreditLines and LoanTerm as numerical first, might be wrong

Model 1:
---
- Linear regression predicting LoanAmount as the Y
- NumCreditLines and LoanTerm and InterestRate treated as numerical
- also excluded default from the model

```python
# get only the numerics in the df_sample
x_numeric = df_sample_1[numeric_1]

# use OLS in scikit learn
model_1 = LinearRegression()
model_1.fit(x_numeric, y)
output = pd.DataFrame({
    'OLS':np.r_[
        model_1.intercept_,
        model_1.coef_
    ]
}, index=["Intercept"]+ numeric_1)
display(output)
```

```python
model_1.score(x_numeric, y)
```
